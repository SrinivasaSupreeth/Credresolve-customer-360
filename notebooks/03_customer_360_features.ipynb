```python
import pandas as pd
```

```python
wa = pd.read_csv("../data/whatsapp_activity.csv")
wa.head()
```

```python
wa['sent_at'] = pd.to_datetime(wa['sent_at'])
wa['read_at'] = pd.to_datetime(wa['read_at'])
```

```python
wa_features = wa.groupby('lead_code').agg(
    wa_msgs_sent=('sent_at', 'count'),
    wa_msgs_read=('status', lambda x: (x == 'READ').sum()),
    wa_msgs_failed=('status', lambda x: (x == 'FAILED').sum()),
    wa_unique_responses=('response_message', 'nunique')
).reset_index()

wa_features.head()
```

```python
bot = pd.read_csv("../data/teleco_call_back.csv")
bot.head()
```

```python
bot_features = bot.groupby('lead_code').agg(
    bot_calls=('start_time', 'count'),
    bot_answered=('disposition', lambda x: (x == 'ANSWERED').sum()),
    bot_avg_duration=('duration', 'mean')
).reset_index()

bot_features.head()
```

```python
human = pd.read_csv("../data/call_placed.csv")
human.head()
```

```python
human_features = human.groupby('lead_code').agg(
    human_calls=('start_time', 'count'),
    human_answered=('disposition', lambda x: (x == 'ANSWERED').sum()),
    human_avg_duration=('duration', 'mean')
).reset_index()

human_features.head()
```

```python
field = pd.read_csv("../data/mobile_app_data.csv")
field.head()
```

```python
field_features = field.groupby('lead_code').agg(
    field_visits=('visit_date', 'count'),
    met_customer=('result', lambda x: (x == 'MET_CUSTOMER').sum()),
    door_locked=('result', lambda x: (x == 'DOOR_LOCKED').sum()),
    shifted=('result', lambda x: (x == 'SHIFTED').sum())
).reset_index()

field_features.head()
```

```python
sms = pd.read_csv("../data/AI_sms_callback.csv")
sms.head()
```

```python
sms_features = sms.groupby('lead_code').agg(
    sms_sent=('status', 'count'),
    sms_failed=('status', lambda x: (x != 'DELIVERED').sum())
).reset_index()

sms_features.head()
```

```python
customer_360 = wa_features \
    .merge(bot_features, on='lead_code', how='outer') \
    .merge(human_features, on='lead_code', how='outer') \
    .merge(field_features, on='lead_code', how='outer') \
    .merge(sms_features, on='lead_code', how='outer')

customer_360.head()
```

```python
customer_360 = customer_360.fillna(0)
customer_360.head()
```

```python
customer_360.to_csv("../data/customer_360_features.csv", index=False)
customer_360 = customer_360.fillna(0)
```
